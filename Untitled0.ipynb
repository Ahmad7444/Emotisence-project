# CROPPING **IMAGE**

since opencv reads image as a numpy array, we can use numpy slicing to crop the image

e.g. cropping from bottom is done as:

```python
cropped_img=image.copy()[:550,:,:]
plt.imshow(cropped_img)
plt.show()
```

lets make a function to crop any image

```python
#top,bottom,left,right are scaled from 0-100
def crop(img,top,bottom,left,right):
  x,y,z=img.shape
  return img[(int)(top*x/100):(int)(bottom*x/100),(int)(left*y/100):(int)(right*y/100)]

plt.imshow(crop(image,40,65,40,100))
plt.show()
```

# RGB colorspace

any image in RGB colorspace uses mixture of red,green,blue colours to represent any image

```python
image_red=image.copy()
image_green=image.copy()
image_blue=image.copy()


image_red[:,:,1:]=0
image_green[:,:,[0,2]]=0
image_blue[:,:,0:2]=0

fig , img=plt.subplots(3,2,figsize=(5,15))

img[0][1].imshow(image_red)
img[1][1].imshow(image_green)
img[2][1].imshow(image_blue)

img[0][1].set_title('Red Channel')
img[1][1].set_title('Green Channel')
img[2][1].set_title('Blue Channel')



img[0][0].imshow(image)
img[1][0].imshow(image)
img[2][0].imshow(image)

img[1][0].set_title("Original")
img[0][0].set_title("Original")
img[2][0].set_title("Original")

plt.show()


```

# HSV colorspace

H for hue-->the type of colour

S for saturation---->colour intensity

V for value---->brightness

**visualisatiion:**
Imagine the HSV color space as a cylinder:

Hue is the angle around the cylinder (0° = red, 120° = green, 240° = blue)(openCV:0-179).

Saturation is the distance from the center (0 = gray, 1 = full color)(openCV:0-255).

Value is the height (0 = black, 1 = full brightness)(openCV:0-255).


```python
image_HSV=cv2.cvtColor(image,cv2.COLOR_RGB2HSV)
plt.imshow(image_HSV)
plt.show()
#the image will be very strange because we are trying to represent HSV image in RGB format
```

```python
image_hue=image_HSV.copy()
image_saturation=image_HSV.copy()
image_value=image_HSV.copy()


image_hue[:,:,1:]=0
image_saturation[:,:,[0,2]]=0
image_value[:,:,0:2]=0

fig , img=plt.subplots(3,2,figsize=(5,15))

img[0][1].imshow(image_hue)
img[1][1].imshow(image_saturation)
img[2][1].imshow(image_value)

img[0][1].set_title('Hue Channel')
img[1][1].set_title('Saturation Channel')
img[2][1].set_title('Value Channel')



img[0][0].imshow(image)
img[1][0].imshow(image)
img[2][0].imshow(image)

img[1][0].set_title("Original")
img[0][0].set_title("Original")
img[2][0].set_title("Original")

plt.show()


```

# IMAGE SMOOTHING

 It removes high-frequency content, like edges, from the image and commonly used image processing operation for reducing the image noise. The process removes high-frequency content such as edges from the input image.

```python
def show_image(title, img, cmap=None):
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB), cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

Adding noise to image

```python
image_noised=image.copy()
rows,cols,channels=image_noised.shape
noise = np.random.normal(0,15,(rows,cols,3)).astype(np.uint8)
image_noised+=noise
fig, ax =plt.subplots(1,2)
ax[0].imshow(image)
ax[1].imshow(image_noised)
plt.show()
```

mage smoothing (or blurring) is used to:

Reduce noise

Remove detail

Help in preprocessing before edge detection or thresholding



Averaging Blur (Box Filter):

Takes the average of all pixels in a kernel (e.g., 5×5) and replaces the center pixel with it

```python
avg_blur = cv2.blur(image, (5, 5))
show_image("Averaging Blur (5x5)", avg_blur)
```

Gaussian Blur:

Uses a weighted average where nearby pixels have more influence — preserves edges better than box blur

```python
gauss_blur = cv2.GaussianBlur(image, (5, 5), 0)
show_image("Gaussian Blur (5x5)", gauss_blur)
```

Median Blur:

 Replaces each pixel with the median of its neighborhood.
Best for removing salt-and-pepper noise.

```python
median_blur = cv2.medianBlur(image, 5)
show_image("Median Blur (5)", median_blur)
```

Bilateral Filter:



Preserves edges while smoothing.
Computationally expensive but useful in applications like cartoonization.


```python
bilateral = cv2.bilateralFilter(image, d=9, sigmaColor=75, sigmaSpace=75)
show_image("Bilateral Filter", bilateral)
```
